```python
#importing the libraries functions to get the data from database
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#loading the csv file data to variable data
data=pd.read_csv('salary_Data.csv')
```

```python
data
```

```python
#To load the years of experience value from data variable to independent variable(input value)
independent= data[["YearsExperience"]]
```

```python
#To load the salary variable value from data variable to dependent variable(output value)
dependent= data[["Salary"]]
```

```python
#To derive the graph based on the input and output
plt.scatter(independent,dependent)
```

```python
#To set the label and font size for years of experience(x)
plt.xlabel('YearsExperience',fontsize=20)
```

```python
#To set the label and font size for years of salary(y)
plt.ylabel('Salary',fontsize=20)
```

```python
#To display the graph based on the dependent and independent variable values
plt.show()
```

```python
#importing train_test_split function from sklearn.model selection library
from sklearn.model_selection import train_test_split
#splitting train and test dataset vlaues
X_train, X_test, y_train, y_test = train_test_split(independent, dependent, test_size=1/3,random_state=0)
```

```python
#viewing y test values
y_test
```

```python
#loading Linear Regression Liberary functions
from sklearn.linear_model import LinearRegression
#assign LinearRegression function to Regressor variable
regressor=LinearRegression()
#model creation using LinearRegression function (Y=mx+c)
regressor.fit(X_train, y_train)
```

```python
#viewing weight value
weight=regressor.coef_
```

```python
#printing the value of weight(slope)
print("weight of the model={}".format(weight))
```

```python
#viewing bais value
bais=regressor.intercept_
```

```python
#printing the value of bais(Starting/Orgin value)
print("bais of the model={}".format(bais))
```

```python
#prediction using test dataset values
y_pred=regressor.predict(X_test)
```

```python
#its a war between actual input and predicted output
y_pred
```

```python
#evoluating the model
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
```

```python
#printing r_score value
r_score
```

```python
#loading the pickle library and Save the model
import pickle
filename='finalized_model.sav'
pickle.dump(regressor,open(filename,'wb'))
```

```python
#load the model and predict the output
loaded_model=pickle.load(open('finalized_model.sav','rb'))
result=loaded_model.predict([[15]])
print(result)
```
