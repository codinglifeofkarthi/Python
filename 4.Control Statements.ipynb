```python
#count vowels in the string
text="Artificial Intelligence"
count=0
for char in text.lower():
    if char in "aeiou":
        count +=1
print("Vowels count in the string ", count)
```

```python
#find the largest number in the list
numbers = [10,100,20,50,110,11,99]
largest=numbers[0]
for num in numbers:
    if num > largest:
        largest=num
print("The largest number in the list ", largest)
```

```python
#find the duplicate number in the list
numbers = [110,100,20,50,110,11,20]
duplicates=[]
for num in numbers:
    if numbers.count(num)>1 and num not in duplicates:
        duplicates.append(num)
print("The duplicate numbers in the list ", duplicates)
```

```python
#find the second largest number in the list
numbers=[10,40,50,80,60,75,90]
distictvalues=list(set(numbers))
distictvalues.sort()
print("The second largest value in the list", distictvalues[-2])
```
