```python
data=(0,1,2,3,'HOPE','AI',4)
numbers=[]
strings=[]
for item in data:
    if isinstance(item,int):
        numbers.append(item)
    elif isinstance(item,str):
        strings.append(item)
result=(tuple(numbers),tuple(strings))
print(result)
```
